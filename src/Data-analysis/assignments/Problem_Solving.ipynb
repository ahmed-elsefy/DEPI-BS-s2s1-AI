```python
class Solution:
    def romanToInt(self, s: str) -> int:
        """Convert Roman numeral string to an integer."""
        roman_map = {
            'I': 1, 'V': 5, 'X': 10, 'L': 50,
            'C': 100, 'D': 500, 'M': 1000
        }
        total = 0
        n = len(s)
        
        for i in range(n):
            current_val = roman_map[s[i]]
            if i + 1 < n and current_val < roman_map[s[i + 1]]:
                total -= current_val
            else:
                total += current_val
                
        return total

solver = Solution()

print(solver.romanToInt("III"))      
print(solver.romanToInt("LVIII"))    
print(solver.romanToInt("MCMXCIV"))  
```

```python
from typing import List

class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        """Binary search to find target index or insertion point in O(log n)."""
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = left + (right - left) // 2
            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
                
        return left

solver = Solution()
print(solver.searchInsert([1, 3, 5, 6], 5))  
print(solver.searchInsert([1, 3, 5, 6], 2))  
print(solver.searchInsert([1, 3, 5, 6], 7))  
```

```python
class Solution:
    def calculate(self, s: str) -> int:
        """Evaluate mathematical expression with +, -, *, / without eval()."""
        current_num = 0
        last_num = 0
        total = 0
        sign = '+'
        n = len(s)
        
        for i, char in enumerate(s):
            if char.isdigit():
                current_num = current_num * 10 + int(char)
                
            if (not char.isdigit() and not char.isspace()) or i == n - 1:
                if sign == '+':
                    total += last_num
                    last_num = current_num
                elif sign == '-':
                    total += last_num
                    last_num = -current_num
                elif sign == '*':
                    last_num = last_num * current_num
                elif sign == '/':
                    last_num = int(last_num / current_num)
                    
                sign = char
                current_num = 0
                
        total += last_num
        return total

solver = Solution()
print(solver.calculate("3+2*2"))       
print(solver.calculate(" 3/2 "))       
print(solver.calculate(" 3+5 / 2 "))   
print(solver.calculate("14-3/2*2+1"))  
```
