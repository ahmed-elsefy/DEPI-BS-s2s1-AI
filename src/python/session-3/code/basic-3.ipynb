```python
def factorial (num: int):
    ''' 

    
    '''
    
    if num < 0:
        raise valueError ("factorial is undefind for negative values ")

    if num == 0 or num == 1:
        return 1

    return num * factorial(num - 1)

```

```python
def is_prime(num: int)-> bool:
    if num < 2:
        return False
    for i in range (2,num):
        if num % i == 0:
            return False
    return True
```

```python
is_prime(10)
```

```python
def common_divisors(num_1 : int ,num_2 : int)-> list[int]:
    limit = min(num_1,num_2)
    divisors = []

    for divisor in range(1, limit + 1):
        if num_1 % divisor == 0 and num_2 % divisor == 0:
            divisors.append(divisor)
    return divisors

common_divisors(20,10)        
```

```python
import string 
import secrets
def gen_secure_pass(length: int = 12) -> str:
    if length <= 0 :
        raise ValueError("password length must be more than zero")
    char = (
        string.ascii_letters +
        string.digits +
        string.punctuation
    )
    return "".join(secrets.choice(char) for _ in range(length))
gen_secure_pass()
```

```python
import random
 
# Predefined responses
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}
 
# Function to get a response based on user input
def get_response(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
   
    return random.choice(responses["default"])
 
# Chatbot function
def chatbot():
    print("Chatbot: Hi! How can I assist you today?")
   
    while True:
        user_input = input("User: ").lower()
        response = get_response(user_input)
        print("Chatbot:", response)
       
        if user_input == "goodbye":
            break
 
# Run the chatbot
chatbot()
 
```
