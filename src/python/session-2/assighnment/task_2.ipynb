```python
def Are_you_playing_banjo(name):

    if name.startswith('R') or name.startswith('r') :
       return f"{name} plays banjo"
    else:
       return f"{name} does not play banjo"


Name = input("Enter your name :")
print(Are_you_playing_banjo(Name))
```
