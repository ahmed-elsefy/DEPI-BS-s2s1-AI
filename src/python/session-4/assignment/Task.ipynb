```python
class Task1:
    """
    Task1 class handles the operation of checking whether a number is even or odd.
    
    Attributes:
        number (int): The number to be checked.
    """
    def __init__(self, number: int):
        """
        Initializes the Task1 object with a given number.
        """
        self.number = number

    def check_even_odd(self) -> str:
        """
        Determines whether the number is even or odd.
        
        Returns:
            str: 'Even' if the number is even, otherwise 'Odd'.
        """
        if self.number % 2 == 0:
            return "Even"
        return "Odd"


t1_obj = Task1(7)
result_1 = t1_obj.check_even_odd()
print(f"Task 1 Result (Number 7): {result_1}")
```
