```python
class Task1:
    """
    Task1 class handles the operation of checking whether a number is even or odd.
    
    Attributes:
        number (int): The number to be checked.
    """
    def __init__(self, number: int):
        """
        Initializes the Task1 object with a given number.
        """
        self.number = number

    def check_even_odd(self) -> str:
        """
        Determines whether the number is even or odd.
        
        Returns:
            str: 'Even' if the number is even, otherwise 'Odd'.
        """
        if self.number % 2 == 0:
            return "Even"
        return "Odd"


t1_obj = Task1(8)
result_1 = t1_obj.check_even_odd()
print(f"Task 1 Result (Number 8): {result_1}")
```

```python
class Task2:
    """
    Task2 class calculates the factorial of a given non-negative integer.
    
    Attributes:
        number (int): The integer whose factorial is to be calculated.
    """
    def __init__(self, number: int):
        """
        Initializes the Task2 object with a given number.
        """
        self.number = number

    def calculate_factorial(self) -> int:
        """
        Calculates the factorial of the stored number.
        
        Returns:
            int: Factorial of the number, or -1 if the number is negative.
        """
        if self.number < 0:
            return -1   
        
        factorial = 1
        for i in range(1, self.number + 1):
            factorial *= i
        return factorial


t2_obj = Task2(7)
result_2 = t2_obj.calculate_factorial()
print(f"Task 2 Result (Factorial of 7): {result_2}")
```

```python
class Task3:
    """
    Task3 class checks if a number is a prime number.
    
    Attributes:
        number (int): The number to be evaluated.
    """
    def __init__(self, number: int):
        """
        Initializes the Task3 object with a given number.
        """
        self.number = number

    def is_prime(self) -> bool:
        """
        Checks if the number is prime.
        
        Returns:
            bool: True if prime, False otherwise.
        """
        if self.number <= 1:
            return False
        
        for i in range(2, int(self.number ** 0.5) + 1):
            if self.number % i == 0:
                return False
        return True


t3_obj = Task3(27)
result_3 = t3_obj.is_prime()
print(f"Task 3 Result (Is 27 Prime?): {result_3}")
```

```python
class Task6:
    """
    Task6 class finds the maximum number in a list of numbers.
    
    Attributes:
        numbers (list): A list of numerical values.
    """
    def __init__(self, numbers: list):
        """
        Initializes the Task6 object with a list of numbers.
        """
        self.numbers = numbers

    def find_max(self):
        """
        Finds and returns the maximum number in the list.
        
        Returns:
            int/float or None: The highest number, or None if list is empty.
        """
        if not self.numbers:
            return None
        
        max_val = self.numbers[0]
        for num in self.numbers:
            if num > max_val:
                max_val = num
        return max_val

t6_obj = Task6([120, 459, 290, 990, 349, 778])
result_6 = t6_obj.find_max()
print(f"Task 6 Result (Max Value): {result_6}")
```

```python
class Task7:
    """
    Task7 class removes duplicate elements from a given list while preserving order.
    
    Attributes:
        items (list): The list containing original elements.
    """
    def __init__(self, items: list):
        """
        Initializes the Task7 object with a list.
        """
        self.items = items

    def remove_duplicates(self) -> list:
        """
        Removes duplicate elements from the stored list.
        
        Returns:
            list: A new list with unique items.
        """
        unique_items = []
        for item in self.items:
            if item not in unique_items:
                unique_items.append(item)
        return unique_items


t7_obj = Task7([11, 22, 22, 33, 44, 44, 55, 11, 66])
result_7 = t7_obj.remove_duplicates()
print(f"Task 7 Result (Unique List): {result_7}")
```

```python
class Task8:
    """
    Task8 class verifies if a word or phrase is a palindrome.
    
    Attributes:
        text (str): The input string to evaluate.
    """
    def __init__(self, text: str):
        """
        Initializes the Task8 object with a string.
        """
        self.text = text

    def is_palindrome(self) -> bool:
        """
        Determines whether the string reads the same forwards and backwards.
        
        Returns:
            bool: True if palindrome, False otherwise.
        """
        cleaned_text = self.text.replace(" ", "").lower()
        return cleaned_text == cleaned_text[::-1]


t8_obj = Task8("Bob")
result_8 = t8_obj.is_palindrome()
print(f"Task 8 Result (Is 'Bob' Palindrome?): {result_8}")
```

```python
class Task9:
    """
    Task9 class generates a Fibonacci sequence up to n terms.
    
    Attributes:
        n_terms (int): The number of terms to generate.
    """
    def __init__(self, n_terms: int):
        """
        Initializes the Task9 object with the term count.
        """
        self.n_terms = n_terms

    def generate_fibonacci(self) -> list:
        """
        Generates the Fibonacci series list.
        
        Returns:
            list: Sequence of Fibonacci numbers up to n_terms.
        """
        if self.n_terms <= 0:
            return []
        elif self.n_terms == 1:
            return [0]
        
        sequence = [0, 1]
        for i in range(2, self.n_terms):
            next_term = sequence[-1] + sequence[-2]
            sequence.append(next_term)
        return sequence

t9_obj = Task9(8)
result_9 = t9_obj.generate_fibonacci()
print(f"Task 9 Result (Fibonacci 8 terms): {result_9}")
```

```python
class BankAccount:
    """Manages a bank account with deposit, withdraw, and balance check."""

    def __init__(self, initial_balance: float = 0.0):
        """Initializes account balance."""
        self.balance = float(initial_balance) if initial_balance >= 0 else 0.0

    def deposit(self, amount: float) -> None:
        """Deposits a positive amount."""
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount: float) -> None:
        """Withdraws amount if funds are sufficient."""
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
        else:
            print("Invalid withdrawal amount or insufficient funds.")

    def check_balance(self) -> float:
        """Displays and returns current balance."""
        print(f"Current balance: {self.balance}")
        return self.balance

acc = BankAccount(100)
acc.deposit(50)
acc.withdraw(30)
acc.check_balance()
```

```python
class Calculator:
    """Handles basic arithmetic operations."""

    def add(self, a: float, b: float) -> float:
        """Returns sum of two numbers."""
        return a + b

    def subtract(self, a: float, b: float) -> float:
        """Returns difference between two numbers."""
        return a - b

    def multiply(self, a: float, b: float) -> float:
        """Returns product of two numbers."""
        return a * b

    def divide(self, a: float, b: float):
        """Returns quotient or error on division by zero."""
        return a / b if b != 0 else "Error! Division by zero."

    def calculator(self) -> None:
        """Executes interactive CLI calculator."""
        print("Welcome to the calculator!\n1: Add\n2: Subtract\n3: Multiply\n4: Divide")
        choice = input("Select operation (1-4): ")
        if choice in ['1', '2', '3', '4']:
            try:
                num1, num2 = float(input("Num 1: ")), float(input("Num 2: "))
            except ValueError:
                print("Invalid input!")
                return
            ops = {'1': self.add, '2': self.subtract, '3': self.multiply, '4': self.divide}
            print(f"Result: {ops[choice](num1, num2)}")
        else:
            print("Invalid operation.")

if __name__ == "__main__":
    calc = Calculator()
    calc.calculator()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    """Abstract base class for animals."""

    def __init__(self, name: str):
        self.name = name

    @abstractmethod
    def make_sound(self) -> str:
        """Returns the sound of the animal."""
        pass

    def describe(self) -> None:
        """Prints general description of the animal."""
        print(f"I am a {self.__class__.__name__}, named {self.name}.")

class Dog(Animal):
    def make_sound(self) -> str:
        """Returns dog barking sound."""
        return "Woof!"

class Cat(Animal):
    def make_sound(self) -> str:
        """Returns cat meow sound."""
        return "Meow!"

class Cow(Animal):
    def make_sound(self) -> str:
        """Returns cow moo sound."""
        return "Moo!"

for a in [Dog("Buddy"), Cat("Kitty"), Cow("Bella")]:
    a.describe()
    print(f"Sound: {a.make_sound()}")
```

```python
class TextFileReader:
    """Reads text files and calculates content statistics."""

    def __init__(self, file_path: str):
        self.file_path = file_path
        self.content = ""

    def read_file(self) -> str:
        """Reads and stores file text."""
        try:
            with open(self.file_path, 'r', encoding='utf-8') as f:
                self.content = f.read()
            return self.content
        except Exception as e:
            return f"Error: {e}"

    def count_lines(self) -> int:
        """Returns line count."""
        return len(self.content.splitlines()) if self.content else 0

    def count_words(self) -> int:
        """Returns word count."""
        return len(self.content.split()) if self.content else 0

    def count_characters(self) -> int:
        """Returns character count."""
        return len(self.content)

    def display_content(self) -> None:
        """Displays full file text."""
        print(self.content)
        
with open("sample_test.txt", "w", encoding="utf-8") as f:
    f.write("Hello World!\nWelcome to AI & OOP Diploma.\nAMIT Learning.")

reader = TextFileReader("sample_test.txt")
reader.read_file()
print("--- File Content ---")
reader.display_content()
print(f"Lines: {reader.count_lines()}")
print(f"Words: {reader.count_words()}")
print(f"Characters: {reader.count_characters()}")        
```

```python
def count_word_frequency(words: list) -> dict:
    """Counts word frequencies from a list."""
    counts = {}
    for w in words:
        counts[w] = counts.get(w, 0) + 1
    return counts

class WordFrequencyCounter:
    """Class to count frequencies of words in a list."""

    def __init__(self, words: list):
        self.words = words

    def get_frequencies(self) -> dict:
        """Returns word count dictionary."""
        return count_word_frequency(self.words)

sample = ["Abdo", "Hi", "Ahmed", "Welcome", "Youssef"]
print(count_word_frequency(sample))
```

```python
def read_txt_file(file_path: str) -> str:
    """Reads a text file with error handling."""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return f.read()
    except Exception as e:
        return f"Error: {e}"

class UserExtractor:
    """Extracts username:password pairs from a file into a dictionary."""

    def __init__(self, file_path: str):
        self.file_path = file_path
        self.usernames = {}

    def extract_usernames(self) -> dict:
        """Parses file lines into a credential dictionary."""
        content = read_txt_file(self.file_path)
        if content.startswith("Error"):
            return {}
        for line in content.splitlines():
            if ":" in line:
                u, p = line.split(":", 1)
                self.usernames[u.strip()] = p.strip()
        return self.usernames
    
with open("users.txt", "w", encoding="utf-8") as f:
    f.write("ahmed:pass123\nsara:secret456\nali:my_pwd789\n")

extractor = UserExtractor("users.txt")
extracted_users = extractor.extract_usernames()
print("Extracted Users:", extracted_users)   
```

```python
class Solution:
    """Solves the Two Sum problem using a hash map."""

    def twoSum(self, nums: list[int], target: int) -> list[int]:
        """
        Finds indices of two numbers that add up to target.

        Returns:
            list[int]: The pair of indices.
        """
        seen = {} 

        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i

        return []

solver = Solution()
print(solver.twoSum([2, 7, 11, 15], 9))  # Output: [0, 1]
print(solver.twoSum([3, 2, 4], 6))  # Output: [1, 2]
print(solver.twoSum([3, 3], 6))  # Output: [0, 1]
```

```python
from typing import Optional


class ListNode:
    """Represents a node in a singly linked list."""

    def __init__(self, val: int = 0, next: Optional["ListNode"] = None):
        """Initializes the node with a value and pointer to the next node."""
        self.val = val
        self.next = next


class Solution:
    """Handles adding two numbers represented by linked lists."""

    def addTwoNumbers(
        self, l1: Optional[ListNode], l2: Optional[ListNode]
    ) -> Optional[ListNode]:
        """
        Adds two numbers stored in reverse order in linked lists.

        Returns:
            ListNode: Head of the resulting sum linked list.
        """
        dummy_head = ListNode(0)
        current = dummy_head
        carry = 0

        while l1 or l2 or carry:
            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            total = val1 + val2 + carry
            carry = total // 10
            current.next = ListNode(total % 10)

            current = current.next
            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next

        return dummy_head.next


def build_linked_list(elements: list) -> Optional[ListNode]:
    """Helper function to create a linked list from a Python list."""
    dummy = ListNode(0)
    curr = dummy
    for x in elements:
        curr.next = ListNode(x)
        curr = curr.next
    return dummy.next


def to_list(node: Optional[ListNode]) -> list:
    """Helper function to convert a linked list into a Python list."""
    result = []
    while node:
        result.append(node.val)
        node = node.next
    return result




solver = Solution()

# Example 1: 342 + 465 = 807
l1 = build_linked_list([2, 4, 3])
l2 = build_linked_list([5, 6, 4])
print("Example 1 Output:", to_list(solver.addTwoNumbers(l1, l2)))  # [7, 0, 8]

# Example 2: 0 + 0 = 0
l1 = build_linked_list([0])
l2 = build_linked_list([0])
print("Example 2 Output:", to_list(solver.addTwoNumbers(l1, l2)))  # [0]

# Example 3: 9999999 + 9999 = 10009998
l1 = build_linked_list([9, 9, 9, 9, 9, 9, 9])
l2 = build_linked_list([9, 9, 9, 9])
print(
    "Example 3 Output:", to_list(solver.addTwoNumbers(l1, l2))
)  # [8, 9, 9, 9, 0, 0, 0, 1]
```
